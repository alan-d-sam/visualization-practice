```python
import matplotlib.pyplot as plt
students = ['Ayush','Alice','Anwar','Michael','Jhon']
marks = [23,45,12,67,100]
```

```python
plt.figure(figsize=(10,5),facecolor='#f6f595') # 10 - width ; 5 - height
plt.bar(students,marks,color='green',alpha=0.5) # alpha changes opacity
plt.xlabel('STUDENT NAME')
plt.ylabel('MARKS')
plt.title('BAR CHART',fontsize=20, fontweight='bold',color='#631100')
plt.show()
```

```python
plt.figure(figsize=(10,5),facecolor='#f6f595') # 10 - width ; 5 - height
plt.plot(students,marks,color='green',linewidth='2',linestyle=':',marker='o',label='Marks') #label = marks
plt.legend(loc="lower right") # show legend when using label, loc - for the location of the legend
plt.xlabel('STUDENT NAME')
plt.ylabel('MARKS')
plt.title('LINE CHART',fontsize=20, fontweight='bold',color='#631100')
plt.grid(True,alpha=0.3)
plt.show()
```

```python
import pandas as pd
data = {
    "year":[2021,2022,2023,2024,2025],
    "Laptop":[100,120,130,140,150],
    "Mobile":[80,75,60,90,80],
    "Tablet":[34,12,13,90,100]
}
df = pd.DataFrame(data)
```

```python
plt.figure(figsize=(10,5),facecolor='#f6f595') # 10 - width ; 5 - height
plt.xlabel('Year')
plt.ylabel('Sales')
plt.xticks(df['year']) # removal of decimal in xlabel
plt.plot(df['year'],df['Laptop'],label='Laptop',color='red',marker='o',linestyle='dotted',ms=5) # ms/marker size
plt.plot(df['year'],df['Mobile'],label='Mobile',color='blue',marker='o',linestyle='dotted',ms=5)
plt.plot(df['year'],df['Tablet'],"go--",label='Tablet',ms=5) # go--
plt.legend()
plt.title('SALES ANALYSIS',fontsize=20, fontweight='bold',color='#631100')
plt.grid(True,alpha=0.3)
plt.show()
```
