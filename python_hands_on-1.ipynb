Q6. Find the factorial of a number

```python
num = int(input("Enter a number: "))
fact = 1
for i in range(num,0,-1):
  fact = i*fact
print(f"Factorial of {num} is {fact}")
```

Q7. Find the sum of digits of a number

```python
num = int(input("Enter a number: "))
sum = 0
while num > 0:
  rev = num%10
  sum += rev
  num = num//10
print(sum)
```
