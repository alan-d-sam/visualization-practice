```python
n = 500
df2 = pd.DataFrame({
    'Age': np.clip(np.random.normal(35, 10, n).astype(int), 18, 70),
    'Annual_Income': np.round(np.random.lognormal(10.8, 0.4, n), 2),
    'Purchase_Amount': np.round(20 + np.random.lognormal(10.8, 0.4, n) * 0.003 + np.random.gamma(2, 25, n), 2),
    'Items_Ordered': np.random.poisson(3, n) + 1,
    'Satisfaction_Score': np.random.choice([1, 2, 3, 4, 5], n, p=[0.1, 0.15, 0.25, 0.35, 0.15]),
    'Segment': np.random.choice(['Standard', 'Premium', 'VIP'], n, p=[0.6, 0.3, 0.1])
})
df2
```

```python
df2.head()
```

```python
sns.boxplot(data=df2, x="Segment", y="Purchase_Amount")


plt.show()
```
