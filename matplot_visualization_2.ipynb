### **Pie Chart**

```python
import pandas as pd

data = {
    "Payment_Method": [
        "UPI",
        "Credit Card",
        "Debit Card",
        "Cash",
        "Net Banking"
    ],

    "Number_of_Customers": [
        45,
        25,
        15,
        8,
        7
    ]
}

df = pd.DataFrame(data)

df
```

```python
plt.figure(figsize=(7, 5))

color = ["red", "blue", "yellow", "orange", "green"]

plt.pie(df["Number_of_Customers"], labels = df["Payment_Method"], autopct= "%1.1f%%", colors = color)

plt.title("Payment Method Distribution")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.pie(df["Number_of_Customers"], labels = df["Payment_Method"], autopct= "%1.1f%%", explode = (0, 0, 0, 0.09, 0))

plt.title("Payment Method Distribution")

plt.show()
```

### **Adding Multiple Plots**

```python
import pandas as pd

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 180, 140, 200, 220],
    "Expenses": [80, 90, 100, 95, 120, 130],
    "Customers": [50, 65, 72, 60, 85, 95],
    "Profit": [40, 60, 80, 45, 80, 90]
}

df = pd.DataFrame(data)

df
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 5))

axes[0, 0].plot(df["Month"], df["Sales"], marker="o", linestyle="--", color="blue")

axes[0, 1].bar(df["Month"], df["Expenses"], color="orange")

axes[1, 0].barh(df["Month"], df["Profit"], color="green")

axes[1, 1].plot(df["Month"], df["Profit"], marker = "P", linestyle="dashdot", color="red")

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 5))

axes[0, 0].plot(df["Month"], df["Sales"], marker="o", linestyle="--", color="blue")
axes[0, 0].set_title("Sales")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Sales")

axes[0, 1].bar(df["Month"], df["Expenses"], color="orange")
axes[0, 1].set_title("Expenses")
axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Expenses")

axes[1, 0].barh(df["Month"], df["Profit"], color="green")
axes[1, 0].set_title("Profit")
axes[1, 0].set_xlabel("Profit")
axes[1, 0].set_ylabel("Month")

axes[1, 1].plot(df["Month"], df["Profit"], marker = "P", linestyle="dashdot", color="red")
axes[1, 1].set_title("Profit")
axes[1, 1].set_xlabel("Month")
axes[1, 1].set_ylabel("Profit")


plt.tight_layout()
plt.show()
```
